#  Исследование результатов А/В-теста и поиск инсайтов

SollmaFin — международное мобильное приложение для онлайн-торговли финансовыми активами. Пользователи могут инвестировать деньги в акции, валюту, криптовалюту, биржевые фонды (ETF) и другие активы. Целевая аудитория приложения — начинающие инвесторы. Получить доступ к финансовым рынкам можно со смартфона либо через веб-версию. 

Приложение ориентировано на рынок Латинской Америки с акцентом на четыре страны: Мексику, Бразилию, Колумбию и Аргентину — там быстро растёт интерес к финтеху и инвестициям, а экономики очень разные. 

Появилась гипотеза о недостаточной финансовой грамотности пользователей: они не понимают, как работать с активами разной степени риска. Клиенты покупают активы с высоким риском и теряют деньги, что снижает вовлечённость.
Команда продукта решила обновить онбординг пользователей и добавить в него детальную информацию о различиях финансовых активов и связанных с ними рисках. Однако появилось опасение, что углублённый онбординг отпугнёт клиентов от пополнения депозита и более рискованных вложений. 

Чтобы разобраться, команда решила провести A/B-эксперимент. 

### А/В-эксперимент

В эксперименте участвовали новые пользователи, которые зарегистрировались в приложении со 2 по 15 июня 2025 года. 
Пользователей случайным образом разделили на две равные группы: 
- Контрольная группа проходила стандартный онбординг без обязательного обучения.
- Тестовая группа проходила обновлённый онбординг с подробной информацией об активах и связанных с ними рисках. 

После разделения активность пользователей анализировали в течение недели.

Цель эксперимента — оценить, как обновлённый онбординг влияет на поведение пользователей и их дальнейшую инвестиционную активность.

При внедрении новой фичи команда продукта выдвинула такие гипотезы:
- Гипотеза роста: обучающий онбординг помогает пользователям лучше понимать принципы инвестирования, поэтому они будут чаще открывать второй депозит.
- Гипотеза риска: информация о возможных потерях и высоких рисках отпугнёт некоторых новичков, особенно самых осторожных, что снизит конверсию в первый депозит.
- Дополнительная гипотеза: после нового онбординга пользователи, которые выбрали высокорискованные активы, будут чаще, чем раньше, возвращаться и открывать второй депозит. При старом онбординге пользователи часто покупали активы с высоким риском без понимания последствий. Это приводило к потерям и оттоку после первого депозита. 

Чтобы увидеть полную картину, недостаточно следить за метриками конверсии. Команда выбрала для эксперимента такой набор метрик: 
- Ключевая метрика — средняя сумма всех депозитов на одного пользователя (включая тех, кто установил приложение или открыл веб-версию).
- Барьерная метрика — конверсия из регистрации в первый депозит.
- Вспомогательная метрика 1 — конверсия из первого депозита во второй.
- Вспомогательная метрика 2 — средняя сумма всех депозитов на пользователя, который открыл хотя бы один депозит.

Ожидалось, что ключевая и барьерная метрики не упадут, а вспомогательные покажут значительный рост.

**Задача 2.3.** Влияние уровня риска актива на открытие второго депозита.
- Изучите, как уровень риска первого актива влияет на конверсию во второй депозит.  

````python
risk_events = ['asset_purchase', 'second_deposit']
events_risk_df = hist_df[hist_df['event_name'].isin(risk_events)]

risk_info = events_risk_df[~events_risk_df['risk_level'].isna()][['user_id','risk_level']]
result = pd.merge(events_risk_df[['user_id', 'event_name']], risk_info, on='user_id', how='left')

result_lst = []

for risk_level in result['risk_level'].unique():
    risk_df = result[result['risk_level']==risk_level]
    risk_df = risk_df.groupby('event_name')['user_id'].nunique().reset_index(name='users_count')
    risk_df['risk_level'] = risk_level
    risk_df['CR'] = round(risk_df['users_count'] / risk_df['users_count'].iloc[0],2)
    result_lst.append(risk_df)

result_df = pd.concat(result_lst)

result_df
````

````python
plot_df = result_df[result_df['event_name'] == 'second_deposit'].copy()

risk_levels = ['low', 'medium','high']
plot_df['risk_level'] = pd.Categorical(
    plot_df['risk_level'],
    categories=risk_levels,
    ordered=True
)
plot_df = plot_df.sort_values('risk_level')

plt.figure(figsize=(10, 6))
bars = plt.bar(plot_df['risk_level'], plot_df['CR'])

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f'{float(height)}',
        ha='center',
        va='bottom'
    )

plt.title('Количество пользователей на second_deposit по risk_level')
plt.xlabel('Risk level')
plt.ylabel('Число пользователей')
plt.tight_layout()
plt.show()
````

Анализ показывает что уровень риска опеределенно влияет на конверсию во второй депозит. Вероятно пользователи покупающие рискованные активы чаще сталкиваются с потерей средств что негативно сказывается на конверсии во второй депозит

**Задача 2.4.** Анализ ключевой метрики на исторических данных.
- На исторических данных рассчитайте значение ключевой метрики — средней суммы всех депозитов на одного пользователя. Учитывайте всех пользователей, которые взаимодействовали с сервисом. Это позволит зафиксировать базовый уровень монетизации.
- Визуализируйте данные, где необходимо. По результатам исследовательского анализа сделайте краткий вывод. 

````python
hist_df['event_date'] = hist_df['event_ts'].dt.date
result_df = (
    hist_df.groupby(hist_df['event_date'])
    .agg(
        total_amount=('amount', 'sum'),
        users_count=('user_id', 'nunique')
    )
    .reset_index()
    .rename(columns={'event_date': 'date'})
)

result_df['key_metric'] = result_df['total_amount'] / result_df['users_count']
````

````python
plt.figure(figsize=(12, 8))

plt.plot(result_df['date'], result_df['key_metric'])
plt.title('Динамика ключевой метрики')
plt.xlabel('Дата')
plt.ylabel('Ключевая метрика')
plt.grid(True)

ax = plt.gca()
ax.xaxis.set_major_locator(mdates.DayLocator(interval=3))   # каждый 3 день
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

На графике виден очень стремительный рост ключевой метрики с начала лета

````python
abt_df['event_date'] = abt_df['event_ts'].dt.floor('D')

daily_arpu = []

for current_date in sorted(abt_df['event_date'].dropna().unique()):
    event_to_date = abt_df[abt_df['event_date'] <= current_date]

    daily_arpu_per_group = (
        event_to_date.groupby('group')
        .agg(
            total_amount=('amount', 'sum'),
            users_count=('user_id', 'nunique')
        )
        .reset_index()
    )

    daily_arpu_per_group['cum_arpu'] = (
        daily_arpu_per_group['total_amount'] / daily_arpu_per_group['users_count']
    )
    daily_arpu_per_group['date'] = current_date

    daily_arpu.append(daily_arpu_per_group[['group', 'date', 'cum_arpu']])

arpu_dynamics_df = pd.concat(daily_arpu, ignore_index=True)

plt.figure(figsize=(12, 8))

for group in arpu_dynamics_df['group'].unique():
    plot_df = arpu_dynamics_df[arpu_dynamics_df['group'] == group]
    plt.plot(plot_df['date'], plot_df['cum_arpu'], label=group)

plt.title('Кумулятивный ARPU по группам')
plt.xlabel('Дата')
plt.ylabel('Кумулятивный ARPU')
plt.grid(True)
plt.legend()

ax = plt.gca()
ax.xaxis.set_major_locator(mdates.DayLocator(interval=1))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

Тест показывает умеренный положительный эффект по ARPU, однако разрыв между группами невелик и требует проверки на статистическую значимость. Также на графике виден скачок в начале лета и это подтверждается графиками динамики ключевой метрики по группам (см. предыдущий график)

````python
abt_df['event_date'] = pd.to_datetime(abt_df['event_ts']).dt.date

daily_user_revenue = (
    abt_df.groupby(['event_date', 'user_id', 'group'], as_index=False)
    .agg(daily_amount=('amount', 'sum'))
)

days = sorted(daily_user_revenue['event_date'].dropna().unique())

pvalue_dynamics = []

for day in days:
    cumulative_df = daily_user_revenue[daily_user_revenue['event_date'] <= day]
    
    user_revenue = (
        cumulative_df.groupby(['user_id', 'group'], as_index=False)
        .agg(total_amount=('daily_amount', 'sum'))
    )
    
    control = user_revenue.loc[user_revenue['group'] == 'control', 'total_amount']
    test = user_revenue.loc[user_revenue['group'] == 'test', 'total_amount']
    
    _, p_value = ttest_ind(
        control,
        test,
        equal_var=False,
        alternative='less'
    )
    
    pvalue_dynamics.append({
        'date': day,
        'p_value': p_value
    })

pvalue_df = pd.DataFrame(pvalue_dynamics)

plt.figure(figsize=(12, 8))
plt.plot(pvalue_df['date'], pvalue_df['p_value'], marker='o')
plt.axhline(0.05, linestyle='--')

plt.title('Динамика p-value по ключевой метрике')
plt.xlabel('Дата')
plt.ylabel('p-value')
plt.grid(True)

ax = plt.gca()
ax.xaxis.set_major_locator(mdates.DayLocator(interval=1))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

Несмотря на тренд к снижению p-value, статистически значимого эффекта не получено - на текущих данных нельзя подтвердить влияние теста.

**Барьерная метрика**

````python
registration = abt_df[abt_df['event_name'] == 'registration']

first_deposit = abt_df[
    (abt_df['event_name'] == 'first_deposit') &
    (abt_df['user_id'].isin(registration['user_id']))
]

reg_count = registration.groupby('group')['user_id'].nunique()
first_dep_count = first_deposit.groupby('group')['user_id'].nunique()

results_df = pd.DataFrame({
    'registration': reg_count,
    'first_deposit': first_dep_count
}).fillna(0)

results_df['registration'] = results_df['registration'].astype(int)
results_df['first_deposit'] = results_df['first_deposit'].astype(int)

results_df['CR'] = (
    results_df['first_deposit'] / results_df['registration'] * 100
).round(2)

display(results_df)
````

Сформулируем гипотезы для проверки:
- Конверсия из регистрации в первый депозит для контрольной и тестовой группы неизменна.
        H_0: p_control = p_test   
    
- Конверсия из регистрации в первый депозит для тестовой группы < , чем для контрольной.
        H_1: p_control > p_test 

<div class="alert alert-info"> <b>Комментарий:</b> Добавил гиоптезы.</div>

````python
control_cr = results_df.loc['control', 'CR']
test_cr = results_df.loc['test', 'CR']

success = [
    results_df.loc['control', 'first_deposit'],
    results_df.loc['test', 'first_deposit']
]

nobs = [
    results_df.loc['control', 'registration'],
    results_df.loc['test', 'registration']
]

stat, p_value = proportions_ztest(count=success, nobs=nobs, alternative='larger')

diff_df = pd.DataFrame({
    'control_CR': [control_cr],
    'test_CR': [test_cr],
    'abs_diff': [test_cr - control_cr],
    'rel_diff_pct': [(test_cr - control_cr) / control_cr * 100],
    'p_value': [p_value]
}).round(2)

display(diff_df)
````

Значение p_value(0.06) > 0.05. Таким образом данных не достаточно чтобы отвергнуть H0 в пользу H1 и утрверждать что барьерная метрка имеет статистически значимое изменение

````python
control_cr = results_df.loc['control', 'CR']
test_cr = results_df.loc['test', 'CR']

success = [
    results_df.loc['control', 'second_deposit'],
    results_df.loc['test', 'second_deposit']
]

nobs = [
    results_df.loc['control', 'first_deposit'],
    results_df.loc['test', 'first_deposit']
]

stat, p_value = proportions_ztest(count=success, nobs=nobs, alternative='smaller')

diff_df = pd.DataFrame({
    'control_CR': [control_cr],
    'test_CR': [test_cr],
    'abs_diff': [test_cr - control_cr],
    'rel_diff_pct': [(test_cr - control_cr) / control_cr * 100],
    'p_value': [p_value]
})

display(diff_df)
````

С точки зрения бизнеса это означает смещение модели монетизации в сторону high-value пользователей. Такой эффект может быть выгоден при ориентации на рост LTV, однако он несет риски, связанные со снижением доли массовых пользователей и потенциальным ухудшением конверсии на входе. Также возникает зависимость выручки от более узкого сегмента аудитории.

В текущем виде фича выглядит перспективной, но требует осторожного внедрения. Рекомендуется продолжить эксперимент для повышения статистической мощности, а также рассмотреть частичный rollout на сегменты, где эффект наиболее выражен (например, high-risk пользователи). Дополнительно важно проработать первый шаг воронки, чтобы минимизировать негативное влияние на конверсию. В целом, фича улучшает монетизацию и поведение платящих пользователей, но сопровождается рисками, которые необходимо учитывать перед масштабированием.
